### INFORMATION ABOUT THE FEATURES-
* id-an unique ID assigned for the loan listing

* member id- an unique ID assigned for the borrower member

* loan_amnt-the amount that was applied for.

* funded_amnt-the amount that was lended.

* funded_amnt_inv-the amount that was invested exc lending club.

* term - duration of loan 36/60 months

* int_rate - interest rate

* installment-the monthly payment owed by borrower

* grade- loan assigned loan grade
(from A to G)

* subgrade - loan assigned subgrade(A1,A2,A3,A4,A5 and so on till G)

* emp_title - job title provided by the borrower

* home_ownership- home status of borrower

* annual_inc - annual income of borrower

* verification_status - if borrowers income was verified or not

* issue_d - issue date of loan

* **loan_status - the current status of loan**


  *   Fully Paid - completely paided off
*   Charged Off - DEFAULTED
*   Current - The loan is active
*   Late - 31 to 120 days late
*   Late - 16 to 30 days late
*   In grace period - just missed the due date and is late by few days
*   Does not meet the credit policy. Status:Fully Paid - Fully Paid
*   Does not meet the credit policy. Status:Charged Off - Charged Off
*   Defaulted - defaulted

* purpose - purpose stated for the loan

* title - provided by the borrower

* zip_code - geographical use

* addr_state - the state of borrower

*  **dti - the debt to income ratio (HIGHER MEANS MORE CHANCES OF BEING A DEFAULTER)**

* delinq_2yrs - how many times the borrower has missed a 30+days due for the loan in last 2 years.

* earliest_cr_line - The month the borrower’s earliest reported credit line was opened.
    *   Helps assess the length of the borrower’s credit history.

    *   A longer credit history is generally a sign of lower credit risk (more experience managing debt).

* inq_last_6mths - Number of credit inquiries in the last 6 months (prior to loan application).This value indicates recent borrowing activity or attempts to get new credit.

* mths_since_last_delinq - The number of months since the borrower's last delinquency

* mths_since_last_record- The number of months since the borrower’s last public record.
A public record on a credit report usually refers to:

  *  Bankruptcy

  *  Tax lien

  *  Judgment

  These are serious negative marks and indicate severe credit events.

* open_acc - The number of open credit lines in the borrower's credit file.average value is considered good.

* pub_rec - indicates the number of derogatory public records listed on the borrower’s credit report.Public records are strong indicators of prior financial distress.Even one public record can significantly increase the likelihood of default.

* revol_bal - represents the total outstanding balance on all revolving credit accounts.A high revol_bal means the borrower is carrying large amounts of credit card or line-of-credit debt, which is unsecured and high-interest.

* revol_util - the percentage of available revolving credit (e.g., credit cards, lines of credit) that the borrower is currently using.

  *  revol_util = revol_bal/credit limit * 100

* total_acc	- total credit accounts

* initial_list_status -  indicates the initial status of the loan when it was first listed on the LendingClub platform.

  * w	(Whole)	- The loan was available for a single investor to fund in full. Typically institutional investors.
  * f	(Fractional) -	The loan was listed for multiple retail investors to fund in parts (i.e., "notes").

* out_prncp -  the total unpaid principal amount that is still outstanding on the loan at the time of record.
It does not include any interest — only the remaining principal that the borrower still owes.

* total_pymnt - the cumulative amount of money the borrower has paid to date for the loan — including:
  * Principal
  * Interest
  * Fees
  * Late fees (if any)

  It reflects the actual cash flow received from the borrower over the life of the loan (until the current date or charge-off).

* total_rec_prncp - refers to the total amount of principal (not interest) that has been repaid by the borrower to date.

  It excludes:

  * Interest payments

  * Late fees

  * Service charges

  It tells us how much of the original loan amount the borrower has actually paid back.

* total_rec_int - the total amount of interest that the borrower has paid on the loan up to the current reporting date.

* total_rec_late_fee - the cumulative amount of late fees that the borrower has actually paid on the loan till date.

* recoveries - the amount of money recovered from the borrower after a loan was charged-off (i.e., marked as a loss).
It only applies to defaulted loans and indicates post-default cash recovery.

* collection_recovery_fee - the amount paid by LendingClub (and charged to investors) as a fee for recovering funds from borrowers after a loan was charged-off.It represents the cost of hiring collection agencies, legal services, or internal recovery efforts.

* last_pymnt_d - last payment date(month)

* last_pymnt_amnt	- last payment amount(principal + interest + any fee).

* next_pymnt_d - next payment date

* last_credit_pull_d - The last date LendingClub pulled the borrower’s credit report from a bureau (e.g., TransUnion).

* collections_12_mths_ex_med - The number of collection accounts (excluding medical) opened in the borrower’s credit report in the last 12 months.Reflects recent delinquency behavior, excluding medical collections (which are less predictive of credit risk).

* mths_since_last_major_derog - The number of months since the borrower’s last major derogatory mark, such as:

* policy_code - This code indicates the credit policy version under which the loan was issued.

* application_type - indicates whether the loan application was made by a single borrower or jointly with a co-borrower.

*  acc_now_delinq - The borrower currently has 2 accounts that are delinquent (i.e., past due and unpaid).

* tot_coll_amt - The total amount currently in collections, excluding medical, across all of the borrower’s accounts.Represents unpaid debt sent to collections. A high value implies ongoing credit distress.

* tot_cur_bal - The sum of the current outstanding balances on all open credit lines (e.g., credit cards, mortgages, auto loans).

* total_rev_hi_lim - The sum of all revolving credit limits (mostly credit cards) that the borrower has available, based on their credit report.


* good_loan -
    * 1 represents paid off
    * 0 represents defaulter
